```python
# ============================================================
# Assignment: Upper Confidence Bound (UCB)
# Multi-Armed Bandit Problem
# ============================================================

import math

# ------------------------------------------------------------
# Part A - Machine Initialization
# ------------------------------------------------------------

machines = ['A', 'B', 'C']

M = 3              # Number of machines
N_play = 10        # Number of plays

# N[a] = number of times machine a has been selected
N = [0, 0, 0]

# R[a] = total reward obtained from machine a
R = [0, 0, 0]

# Q[a] = estimated value of machine a
Q = [0.0, 0.0, 0.0]

# UCB[a] = Upper Confidence Bound value
UCB = [float('inf'), float('inf'), float('inf')]

# Total reward received
total_reward = 0


# ------------------------------------------------------------
# Function to calculate UCB
# ------------------------------------------------------------

def calculate_ucb():
    ucb_values = []

    total_selections = sum(N)

    for i in range(M):

        # Machines that have never been selected
        # are given infinity so they are selected first.
        if N[i] == 0:
            ucb_values.append(float('inf'))

        else:
            confidence = math.sqrt(
                (2 * math.log(total_selections)) / N[i]
            )

            ucb = Q[i] + confidence
            ucb_values.append(ucb)

    return ucb_values


# ------------------------------------------------------------
# Part C/D/E - Play the machines
# ------------------------------------------------------------

print("=" * 70)
print("        UPPER CONFIDENCE BOUND - MULTI-ARMED BANDIT")
print("=" * 70)

print("\nMachines:", machines)
print("Number of plays:", N_play)

print("\nEnter the reward obtained whenever a machine is selected.")
print("Suggested reward patterns:")
print("A -> 3, 4, 2, 5")
print("B -> 6, 7, 5, 8")
print("C -> 4, 3, 5, 4")

print("\n" + "-" * 70)


for play in range(1, N_play + 1):

    # --------------------------------------------------------
    # Part C - Calculate UCB
    # --------------------------------------------------------

    UCB = calculate_ucb()

    # Select machine having highest UCB
    selected_index = UCB.index(max(UCB))
    selected_machine = machines[selected_index]

    print(f"\nPLAY {play}")
    print("-" * 40)

    print("UCB values:")

    for i in range(M):
        if math.isinf(UCB[i]):
            print(f"Machine {machines[i]} : Infinity")
        else:
            print(f"Machine {machines[i]} : {UCB[i]:.4f}")

    print(f"\nSelected Machine: {selected_machine}")

    # --------------------------------------------------------
    # Part D - Reward Collection
    # --------------------------------------------------------

    while True:
        try:
            reward = float(
                input(f"Enter reward obtained from machine {selected_machine}: ")
            )
            break
        except ValueError:
            print("Please enter a valid numerical reward.")

    # --------------------------------------------------------
    # Update N
    # --------------------------------------------------------

    N[selected_index] += 1

    # --------------------------------------------------------
    # Update R
    # --------------------------------------------------------

    R[selected_index] += reward

    # --------------------------------------------------------
    # Part E - Incremental Sample-Average Q Update
    #
    # Q(a) = Q(a) + [R - Q(a)] / N(a)
    # --------------------------------------------------------

    Q[selected_index] = (
        Q[selected_index]
        + (reward - Q[selected_index]) / N[selected_index]
    )

    # Update total reward
    total_reward += reward

    print(f"Reward received: {reward}")
    print(f"Updated N({selected_machine}) = {N[selected_index]}")
    print(f"Updated R({selected_machine}) = {R[selected_index]:.2f}")
    print(f"Updated Q({selected_machine}) = {Q[selected_index]:.4f}")


# ------------------------------------------------------------
# Part F - Final UCB Calculation
# ------------------------------------------------------------

UCB = calculate_ucb()


# ------------------------------------------------------------
# Display Final Results
# ------------------------------------------------------------

print("\n\n")
print("=" * 70)
print("                    FINAL RESULTS")
print("=" * 70)

print(
    f"{'Machine':<12}"
    f"{'Count N':<15}"
    f"{'Reward R':<15}"
    f"{'UCB':<15}"
    f"{'Q-Value':<15}"
)

print("-" * 70)

for i in range(M):

    if math.isinf(UCB[i]):
        ucb_display = "Infinity"
    else:
        ucb_display = f"{UCB[i]:.4f}"

    print(
        f"{machines[i]:<12}"
        f"{N[i]:<15}"
        f"{R[i]:<15.2f}"
        f"{ucb_display:<15}"
        f"{Q[i]:<15.4f}"
    )

print("-" * 70)

print(f"\nTotal Reward Received = {total_reward:.2f}")

# Best machine according to estimated Q-value
best_machine_index = Q.index(max(Q))

print(
    f"Best Machine based on Q-value = "
    f"{machines[best_machine_index]}"
)

print(
    f"Highest estimated Q-value = "
    f"{Q[best_machine_index]:.4f}"
)

print("\nProgram completed successfully.")
```
