```python
# Greedy Multi-Armed Bandit using Arithmetic Mean

n = int(input("Enter number of machines: "))

machines = []
all_rewards = []
q_values = []

for i in range(n):
    machine = input(f"\nEnter machine {i + 1} name: ")
    rewards = list(map(int, input("Enter observed rewards: ").split()))

    q = sum(rewards) / len(rewards)

    machines.append(machine)
    all_rewards.append(rewards)
    q_values.append(q)

print("\n========== RESULTS ==========")

for i in range(n):
    print(f"Machine {machines[i]}")
    print(f"Rewards: {all_rewards[i]}")
    print(f"Q({machines[i]}) = {q_values[i]:.2f}")

best_index = q_values.index(max(q_values))

print("\nGreedy Selected Machine:", machines[best_index])
```

```python
# Greedy Multi-Armed Bandit
# Incremental Update Method

n = int(input("Enter number of machines: "))

machines = []
all_rewards = []
N = []
Q = []
R = []

# Initialize machines and values
for i in range(n):
    machine = input(f"\nEnter machine {i + 1} name: ")
    rewards = list(map(int, input("Enter observed rewards: ").split()))

    machines.append(machine)
    all_rewards.append(rewards)

    N.append(0)
    Q.append(0.0)
    R.append(0)

# Update Q incrementally for every reward
for i in range(n):
    for reward in all_rewards[i]:

        # Increment count
        N[i] += 1

        # Store total reward
        R[i] += reward

        # Incremental update formula
        Q[i] = Q[i] + (reward - Q[i]) / N[i]

# Display results
print("\n========== RESULTS ==========")

for i in range(n):
    print(f"\nMachine {machines[i]}")
    print(f"Rewards: {all_rewards[i]}")
    print(f"N({machines[i]}) = {N[i]}")
    print(f"R({machines[i]}) = {R[i]}")
    print(f"Q({machines[i]}) = {Q[i]:.2f}")

# Find machine with highest Q
best = Q.index(max(Q))

print("\nGreedy Selected Machine:", machines[best])
```
