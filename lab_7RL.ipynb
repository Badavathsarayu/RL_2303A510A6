```python
# Assignment 1: Calculate Undiscounted and Discounted Returns

n = int(input("Enter the number of time steps: "))

rewards = []

for i in range(n):
    reward = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(reward)

gamma = float(input("Enter the discount factor (gamma): "))

# Calculate undiscounted return
undiscounted_return = sum(rewards)

# Calculate discounted return
discounted_return = 0.0

for t in range(n):
    discounted_return += (gamma ** t) * rewards[t]

# Display results
print("\nEntered reward values:", rewards)
print("Undiscounted return:", undiscounted_return)
print("Discounted return:", discounted_return)
```

```python
# Deterministic Grid World MDP

class GridWorld:
    def __init__(self):
        # State Space
        self.states = {
            "S0": (0, 0),
            "S1": (0, 1),
            "G": (0, 2),
            "S3": (1, 1),
            "S4": (1, 2)
        }

        # Action Space
        self.actions = ["Up", "Down", "Left", "Right"]

        # Discount Factor
        self.gamma = 0.9

        # Rewards
        self.normal_reward = -1
        self.goal_reward = 10
        self.obstacle_reward = -5

        # Transition Function
        self.transitions = {
            "S0": {
                "Up": "S0", "Down": "S0",
                "Left": "S0", "Right": "S1"
            },
            "S1": {
                "Up": "S1", "Down": "S3",
                "Left": "S0", "Right": "G"
            },
            "S3": {
                "Up": "S1", "Down": "S3",
                "Left": "S3", "Right": "S4"
            },
            "S4": {
                "Up": "G", "Down": "S4",
                "Left": "S3", "Right": "S4"
            },
            "G": {
                "Up": "G", "Down": "G",
                "Left": "G", "Right": "G"
            }
        }

    # Transition Function
    def transition(self, state, action):
        if state not in self.states:
            raise ValueError("Invalid state")

        if action not in self.actions:
            raise ValueError("Invalid action")

        if state == "G":
            return "G", 0, True

        row, col = self.states[state]

        moves = {
            "Up": (-1, 0),
            "Down": (1, 0),
            "Left": (0, -1),
            "Right": (0, 1)
        }

        dr, dc = moves[action]
        new_position = (row + dr, col + dc)

        # Obstacle at (1, 0)
        if new_position == (1, 0):
            return state, self.obstacle_reward, True

        next_state = self.transitions[state][action]

        # Goal reached
        if next_state == "G":
            return "G", self.goal_reward, True

        return next_state, self.normal_reward, False

    # Estimate Next State
    def get_next_state(self, state, action):
        next_state, reward, done = self.transition(state, action)
        return next_state


if __name__ == "__main__":
    env = GridWorld()

    print("State Space:", list(env.states.keys()))
    print("Action Space:", env.actions)
    print("Discount Factor:", env.gamma)

    state = input("Enter current state (S0, S1, S3, S4, G): ")
    action = input("Enter action (Up, Down, Left, Right): ")

    next_state, reward, done = env.transition(state, action)

    print("Current State:", state)
    print("Action:", action)
    print("Next State:", next_state)
    print("Reward:", reward)
    print("Episode Terminated:", done)
```
