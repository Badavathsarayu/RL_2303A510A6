```python
machines = ["A", "B", "C"]

N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]

plays = int(input("Enter number of plays: "))

for i in range(plays):

    print("\nPlay", i + 1)

    machine = input("Enter machine (A/B/C): ").upper()

    if machine == "A":
        index = 0
    elif machine == "B":
        index = 1
    elif machine == "C":
        index = 2
    else:
        print("Invalid machine!")
        continue

    reward = float(input("Enter reward: "))

    # Increment number of times machine is played
    N[index] += 1

    # Add reward
    R[index] += reward

    # Incremental update formula
    Q[index] = Q[index] + (reward - Q[index]) / N[index]

    # Find machine with highest Q
    best = 0

    for j in range(1, 3):
        if Q[j] > Q[best]:
            best = j

    print("Q(A) =", round(Q[0], 2))
    print("Q(B) =", round(Q[1], 2))
    print("Q(C) =", round(Q[2], 2))

    print("Greedy selects: Machine", machines[best])


print("\n========== FINAL RESULTS ==========")
print("Machine\tN\tR\tQ")

for i in range(3):
    print(
        machines[i],
        "\t", N[i],
        "\t", round(R[i], 2),
        "\t", round(Q[i], 2)
    )
```
