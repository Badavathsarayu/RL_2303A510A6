```python



# Discount Factor
gamma = 0.9

# Initialization

def env_statevalues():
    V = {
        "S0": 0,
        "S1": 0,
        "goal": 0
    }
    return V

V = env_statevalues()

theta = 0.001
max_iter = 1000

states = ["S0", "S1"]
actions = ["up", "down", "right", "left"]

# Transition Definition

transition = {
    "S0": {
        "up": "S0",
        "down": "S1",
        "right": "S0",
        "left": "S0"
    },
    "S1": {
        "up": "S0",
        "down": "S1",
        "right": "goal",
        "left": "S1"
    }
}

# Action Value Initialization

def env_action_value():
    Q = {
        "S0": {"up": 0, "down": 0, "right": 0, "left": 0},
        "S1": {"up": 0, "down": 0, "right": 0, "left": 0}
    }
    return Q

Q = env_action_value()

# Policy Initialization

def env_policy():
    P = {
        "S0": "0",
        "S1": "0"
    }
    return P

P = env_policy()

# Reward Definition
rewards = {
    "goal": 100,
    "wall": -1,
    "move": -1
}

# Next State Estimation

def env_nextstate(state, action):
    next_state = transition[state][action]
    return next_state

# Reward Estimation

def env_reward(next_state):
    if next_state == "goal":
        reward = rewards["goal"]
    elif next_state == "wall":
        reward = rewards["wall"]
    else:
        reward = rewards["move"]
    return reward

# Value Iteration Algorithm

for iter in range(max_iter):

    old_V = V.copy()

    for state in states:

        for action in actions:

            next_state = env_nextstate(state, action)
            reward = env_reward(next_state)

            Q[state][action] = reward + gamma * V[next_state]

        V[state] = max(Q[state].values())

        P[state] = max(Q[state], key=Q[state].get)

    # Calculate maximum difference
    max_diff = max(abs(old_V[s] - V[s]) for s in states)

    print(f"Iteration {iter + 1}: {V}")

    if max_diff < theta:
        break

# Final Results

print("\nFinal State Values =", V)
print("Final Policy =", P)

print("\nFinal Q Values:")
for state in states:
    print(state, Q[state])
```
