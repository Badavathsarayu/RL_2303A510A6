```python
import gymnasium as gym

# Function for Random Agent
def random_agent(env_name):
    env = gym.make(env_name)

    total_rewards = []

    for episode in range(5):
        # 1. Reset the environment
        state, info = env.reset()

        total_reward = 0
        done = False

        print("\nEpisode:", episode + 1)

        while not done:

            # 2. Identify available actions
            # 3. Select a random action
            action = env.action_space.sample()

            # 4. Execute action
            next_state, reward, terminated, truncated, info = env.step(action)

            # 5. Observe next state and reward
            print("State:", state)
            print("Action:", action)
            print("Reward:", reward)
            print("Next State:", next_state)

            # Add reward
            total_reward += reward

            # Move to next state
            state = next_state

            # 6. Check episode termination
            done = terminated or truncated

        # Total reward of episode
        print("Total Reward:", total_reward)
        total_rewards.append(total_reward)

    # 7. Calculate average reward
    average_reward = sum(total_rewards) / len(total_rewards)

    print("\nRewards:", total_rewards)
    print("Average Reward:", average_reward)

    env.close()


# -------------------------------
# 1. CartPole
# -------------------------------
random_agent("CartPole-v1")

# -------------------------------
# 2. MountainCar
# -------------------------------
random_agent("MountainCar-v0")

# -------------------------------
# 3. FrozenLake
# -------------------------------
random_agent("FrozenLake-v1")
```
