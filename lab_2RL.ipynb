```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2

# Part B - Initialize N, R and Q
N = {'A': 0, 'B': 0, 'C': 0}   # Participation count
R = {'A': 0, 'B': 0, 'C': 0}   # Total reward
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}  # Estimated value

# Part C, D - Epsilon-Greedy and Q-value update
for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    if r < epsilon:
        # Exploration
        selected = random.choice(machines)
        decision = "Exploration"
    else:
        # Exploitation
        max_q = max(Q.values())

        # If multiple machines have the same Q value,
        # randomly choose one among them
        best_machines = [m for m in machines if Q[m] == max_q]
        selected = random.choice(best_machines)
        decision = "Exploitation"

    print(f"\nPlay {play}")
    print(f"Random number = {r:.4f}")
    print(f"{decision}: Machine {selected}")

    # Part C - Get reward from user
    reward = float(input(f"Enter reward for machine {selected}: "))

    # Part D - Update count
    N[selected] += 1

    # Update total reward
    R[selected] += reward

    # Incremental sample-average formula:
    # Q_new = Q_old + (Reward - Q_old) / N
    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]

    print(f"Updated Q({selected}) = {Q[selected]:.2f}")


# Part E - Display final results
print("\n========== FINAL RESULTS ==========")

print("\nMachine\tN (Count)\tR (Total Reward)\tQ (Estimated Value)")

for machine in machines:
    print(
        f"{machine}\t{N[machine]}\t\t"
        f"{R[machine]:.2f}\t\t\t"
        f"{Q[machine]:.2f}"
    )



```
