```python
# Assignment 8
# Reinforcement Learning
# Calculate State Value and Action Value
# Without using Bellman Equation

# Discount factor
GAMMA = 0.9

# Rewards
NORMAL_REWARD = -1
GOAL_REWARD = 10
OBSTACLE_REWARD = -5

# States
states = ["S0", "S1", "S3", "S4", "G"]

# Policy given in the assignment
policy = {
    "S0": "right",
    "S1": "down",
    "S3": "right",
    "S4": "up"
}


# ---------------------------------------------------------
# Transition function
# ---------------------------------------------------------
def transition(state, action):
    """
    Returns:
        next_state, reward, done
    """

    # S0 = (0,0)
    if state == "S0":
        if action == "right":
            return "S1", NORMAL_REWARD, False

        elif action == "up":
            # Cannot move outside grid.
            # Agent remains in S0.
            return "S0", NORMAL_REWARD, False

        elif action == "down":
            # X is obstacle
            return "S0", OBSTACLE_REWARD, False

        elif action == "left":
            # Outside grid
            return "S0", NORMAL_REWARD, False


    # S1 = (0,1)
    elif state == "S1":
        if action == "down":
            return "S3", NORMAL_REWARD, False

        elif action == "right":
            return "G", GOAL_REWARD, True

        elif action == "up":
            return "S1", NORMAL_REWARD, False

        elif action == "left":
            return "S0", NORMAL_REWARD, False


    # S3 = (1,1)
    elif state == "S3":
        if action == "right":
            return "S4", NORMAL_REWARD, False

        elif action == "up":
            return "S1", NORMAL_REWARD, False

        elif action == "down":
            return "S3", NORMAL_REWARD, False

        elif action == "left":
            return "S3", NORMAL_REWARD, False


    # S4 = (1,2)
    elif state == "S4":
        if action == "up":
            return "G", GOAL_REWARD, True

        elif action == "left":
            return "S3", NORMAL_REWARD, False

        elif action == "right":
            return "S4", NORMAL_REWARD, False

        elif action == "down":
            return "S4", NORMAL_REWARD, False


    # Goal state
    elif state == "G":
        return "G", 0, True


# ---------------------------------------------------------
# State Value Function
# ---------------------------------------------------------
def state_value_function(start_state):
    """
    Calculate V(s) by following the given policy
    and calculating the discounted return.
    """

    state = start_state
    total_return = 0
    discount = 1

    while True:

        # Select action according to policy
        action = policy[state]

        # Perform action
        next_state, reward, done = transition(state, action)

        # Add discounted reward
        total_return += discount * reward

        # Update discount factor
        discount *= GAMMA

        # Stop if goal is reached
        if done:
            break

        state = next_state

    return total_return


# ---------------------------------------------------------
# Action Value Function
# ---------------------------------------------------------
def action_value_function(start_state, first_action):
    """
    Calculate Q(s,a).

    The first action is specified by the user.
    After the first action, the given policy is followed.
    """

    state = start_state
    action = first_action

    total_return = 0
    discount = 1

    while True:

        # Execute action
        next_state, reward, done = transition(state, action)

        # Add discounted reward
        total_return += discount * reward

        # Update discount
        discount *= GAMMA

        # Stop if goal is reached
        if done:
            break

        # Move to next state
        state = next_state

        # From this point onward follow the policy
        action = policy[state]

    return total_return


# ---------------------------------------------------------
# Main Program
# ---------------------------------------------------------
print("==========================================")
print("     REINFORCEMENT LEARNING - ASSIGNMENT 8")
print("==========================================")

# Calculate state value for S0
v_s0 = state_value_function("S0")

print("\nState Value Function")
print("--------------------")
print("V(S0) =", v_s0)


# Calculate action value for S0 and Up
q_s0_up = action_value_function("S0", "up")

print("\nAction Value Function")
print("---------------------")
print("Q(S0, Up) =", q_s0_up)

print("\n==========================================")
```
